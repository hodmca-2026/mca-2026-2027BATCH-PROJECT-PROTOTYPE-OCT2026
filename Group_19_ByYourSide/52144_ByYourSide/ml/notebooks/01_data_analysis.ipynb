```python
import pandas as pd
import os

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
combined_test = pd.read_csv(
    "../data/raw/mental_health_combined_test.csv"
)

feature_engineered = pd.read_csv(
    "../data/raw/mental_heath_feature_engineered.csv"
)

unbalanced = pd.read_csv(
    "../data/raw/mental_heath_unbanlanced.csv"
)

print("Combined Test:", combined_test.shape)
print("Feature Engineered:", feature_engineered.shape)
print("Unbalanced:", unbalanced.shape)
```

```python
# Create a copy of the original dataset
clean_df = unbalanced.copy()

# Find texts that have more than one different label
text_label_counts = clean_df.groupby("text")["status"].nunique()

conflicting_text_list = text_label_counts[
    text_label_counts > 1
].index

# Remove texts with conflicting labels
clean_df = clean_df[
    ~clean_df["text"].isin(conflicting_text_list)
]

# Remove duplicate texts with the same label
clean_df = clean_df.drop_duplicates(
    subset=["text"],
    keep="first"
)

# Keep only ML-relevant columns
clean_df = clean_df[["text", "status"]].copy()

print("Original dataset shape:", unbalanced.shape)
print("Clean dataset shape:", clean_df.shape)

print("\nClass distribution:")
print(clean_df["status"].value_counts())
```

```python
# Get all texts present in the combined test dataset
test_texts = set(combined_test["text"])

# Remove training texts that also appear in the combined test dataset
independent_df = clean_df[
    ~clean_df["text"].isin(test_texts)
].copy()

print("Original clean dataset:", len(clean_df))
print("Overlapping texts removed:", len(clean_df) - len(independent_df))
print("Independent training pool:", len(independent_df))

print("\nClass distribution:")
print(independent_df["status"].value_counts())
```

```python
X = independent_df["text"]
y = independent_df["status"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))

print("\nTraining class distribution:")
print(y_train.value_counts())

print("\nTesting class distribution:")
print(y_test.value_counts())
```

```python
tfidf = TfidfVectorizer(
    stop_words="english",
    max_features=20000,
    ngram_range=(1, 2),
    min_df=2,
    max_df=0.95
)
```

```python
X_train_tfidf = tfidf.fit_transform(X_train)

X_test_tfidf = tfidf.transform(X_test)

print("Training TF-IDF shape:", X_train_tfidf.shape)
print("Testing TF-IDF shape:", X_test_tfidf.shape)
```

```python
from sklearn.linear_model import LogisticRegression

logistic_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

logistic_model.fit(X_train_tfidf, y_train)

print("Logistic Regression training completed! ✅")
```

```python
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

# Make predictions on unseen test data
y_pred_logistic = logistic_model.predict(X_test_tfidf)

# Accuracy
accuracy = accuracy_score(y_test, y_pred_logistic)

print("Logistic Regression Accuracy:", round(accuracy * 100, 2), "%")

# Detailed metrics
print("\nClassification Report:\n")
print(classification_report(y_test, y_pred_logistic))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Create confusion matrix
cm = confusion_matrix(y_test, y_pred_logistic)

# Plot confusion matrix
plt.figure(figsize=(8, 6))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    xticklabels=logistic_model.classes_,
    yticklabels=logistic_model.classes_
)

plt.xlabel("Predicted Label")
plt.ylabel("Actual Label")
plt.title("Logistic Regression - Confusion Matrix")

plt.show()
```

```python
# Prepare full independent training data

X_final = independent_df["text"]
y_final = independent_df["status"]

print("Final training samples:", len(X_final))

print("\nFinal training class distribution:")
print(y_final.value_counts())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Final production TF-IDF vectorizer
final_tfidf = TfidfVectorizer(
    max_features=30000,
    ngram_range=(1, 2),
    stop_words="english",
    sublinear_tf=True
)

# Learn vocabulary from the complete independent training dataset
X_final_tfidf = final_tfidf.fit_transform(X_final)

print("Final TF-IDF training shape:", X_final_tfidf.shape)
print("Vocabulary size:", len(final_tfidf.vocabulary_))
```

```python
from sklearn.linear_model import LogisticRegression

# Final production Logistic Regression model
final_model = LogisticRegression(
    C=1.0,
    class_weight="balanced",
    max_iter=1000,
    random_state=42
)

# Train using the complete independent training dataset
final_model.fit(X_final_tfidf, y_final)

print("Final production model training completed! 🏆✅")
```

```python
from sklearn.metrics import accuracy_score, f1_score, classification_report

# Transform external data using the FINAL vectorizer
X_external_final = final_tfidf.transform(X_external)

# Make predictions
y_pred_final = final_model.predict(X_external_final)

# Calculate metrics
final_external_accuracy = accuracy_score(
    y_external,
    y_pred_final
)

final_external_macro_f1 = f1_score(
    y_external,
    y_pred_final,
    average="macro"
)

final_external_weighted_f1 = f1_score(
    y_external,
    y_pred_final,
    average="weighted"
)

print(f"Final Model External Accuracy: {final_external_accuracy * 100:.2f}%")
print(f"Final Model External Macro F1: {final_external_macro_f1:.4f}")
print(f"Final Model External Weighted F1: {final_external_weighted_f1:.4f}")

print("\nClassification Report:\n")

print(
    classification_report(
        y_external,
        y_pred_final
    )
)
```

```python
import joblib
from pathlib import Path

# Create models directory if it doesn't exist
model_dir = Path("../models")
model_dir.mkdir(parents=True, exist_ok=True)

# Save the trained Logistic Regression model
joblib.dump(
    final_model,
    model_dir / "checkon_text_classifier.joblib"
)

# Save the fitted TF-IDF vectorizer
joblib.dump(
    final_tfidf,
    model_dir / "checkon_tfidf_vectorizer.joblib"
)

print("Model saved successfully! ✅")
print("Vectorizer saved successfully! ✅")

print("\nSaved files:")
for file in model_dir.iterdir():
    print("-", file.name)
```
